```python

class Book:
    def __init__(self, title, author, available=True):
        self.title = title
        self.author = author
        self.available = available

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author, available):
        book = Book(title, author, available)
        self.books.append(book)
        print("Book added successfully.")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.available:
                    book.available = False
                    print("Book borrowed successfully.")
                else:
                    print("Book is already borrowed.")
                return
        print("Book does not exist.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.available:
                    book.available = True
                    print("Book returned successfully.")
                else:
                    print("Book is already available.")
                return
        print("Book does not exist.")

    def display_books(self):
        if not self.books:
            print("No books in the library.")
            return

        for book in self.books:
            status = "Available" if book.available else "Borrowed"
            print(f"{book.title} by {book.author} - {status}")

library = Library()

while True:
    print("\nLIBRARY BOOK MANAGEMENT")
    print("1. Add Books")
    print("2. Borrow Book")
    print("3. Return Book")
    print("4. Display Books")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        try:
            count = int(input("How many books to add? "))

            if count < 1:
                print("Enter at least one book.")
                continue

            for i in range(count):
                print(f"\nBook {i + 1}")
                title = input("Enter title: ").strip()
                author = input("Enter author: ").strip()
                status = input("Is the book available? (yes/no): ").lower()

                if not title or not author:
                    print("Title and author cannot be empty.")
                    continue

                if status not in ["yes", "no"]:
                    print("Invalid availability status.")
                    continue

                library.add_book(title, author, status == "yes")

        except ValueError:
            print("Please enter a valid number.")

    elif choice == "2":
        title = input("Enter book title to borrow: ").strip()
        library.borrow_book(title)

    elif choice == "3":
        title = input("Enter book title to return: ").strip()
        library.return_book(title)

    elif choice == "4":
        library.display_books()

    elif choice == "5":
        print("Exiting Library System.")
        break

    else:
        print("Invalid choice.")
```

```python

class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        regular_hours = min(self.hours_worked, 160)
        overtime_hours = max(self.hours_worked - 160, 0)

        base_salary = regular_hours * self.hourly_rate
        overtime_pay = overtime_hours * self.hourly_rate * 1.5
        total_salary = base_salary + overtime_pay

        return base_salary, overtime_pay, total_salary

employees = []

try:
    count = int(input("Enter number of employees: "))

    if count < 1:
        print("At least one employee is required.")

    else:
        for i in range(count):
            print(f"\nEmployee {i + 1}")

            name = input("Enter employee name: ").strip()
            hourly_rate = float(input("Enter hourly rate: "))
            hours_worked = float(input("Enter hours worked: "))

            if not name:
                print("Employee name cannot be empty.")
                continue

            if hourly_rate < 0 or hours_worked < 0:
                print("Negative wages or hours are not allowed.")
                continue

            employee = Employee(name, hourly_rate, hours_worked)
            employees.append(employee)

        print("\nEMPLOYEE PAYROLL REPORT")

        for employee in employees:
            base, overtime, total = employee.calculate_salary()

            print("\nEmployee:", employee.name)
            print("Hours Worked:", employee.hours_worked)
            print("Base Salary: P", base)
            print("Overtime Pay: P", overtime)
            print("Total Salary: P", total)

except ValueError:
    print("Invalid input. Please enter valid numbers.")
```

```python

class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

class ShoppingCart:
    def __init__(self):
        self.cart = {}

    def add_product(self, product, quantity):
        if quantity <= 0:
            print("Invalid quantity.")
        elif quantity > product.stock:
            print("Not enough stock available.")
        else:
            product.stock -= quantity

            if product.name in self.cart:
                self.cart[product.name][1] += quantity
            else:
                self.cart[product.name] = [product, quantity]

            print("Product added successfully.")

    def remove_product(self, name):
        if name in self.cart:
            product, quantity = self.cart.pop(name)
            product.stock += quantity
            print("Product removed successfully.")
        else:
            print("Product not found in cart.")

    def display_cart(self):
        if not self.cart:
            print("Your cart is empty.")
            return

        total = 0

        print("\nSHOPPING CART")

        for name, item in self.cart.items():
            product, quantity = item
            subtotal = product.price * quantity
            total += subtotal

            print(name, "x", quantity, "- P", subtotal)

        print("Total Price: P", total)

products = [
    Product("Laptop", 35000, 5),
    Product("Mouse", 500, 10),
    Product("Keyboard", 1200, 8),
    Product("Headphones", 1500, 6)
]

cart = ShoppingCart()

while True:
    print("\nONLINE SHOPPING CART")
    print("1. View Products")
    print("2. Add Product")
    print("3. Remove Product")
    print("4. View Cart")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nAVAILABLE PRODUCTS")

        for product in products:
            print(
                product.name,
                "- P", product.price,
                "- Stock:", product.stock
            )

    elif choice == "2":
        name = input("Enter product name: ").strip()

        selected = None

        for product in products:
            if product.name.lower() == name.lower():
                selected = product
                break

        if selected:
            try:
                quantity = int(input("Enter quantity: "))
                cart.add_product(selected, quantity)
            except ValueError:
                print("Please enter a valid quantity.")
        else:
            print("Invalid product selection.")

    elif choice == "3":
        name = input("Enter product name to remove: ").strip()
        cart.remove_product(name)

    elif choice == "4":
        cart.display_cart()

    elif choice == "5":
        print("Thank you for shopping!")
        break

    else:
        print("Invalid choice.")
```
